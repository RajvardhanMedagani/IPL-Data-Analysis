# IPL Performance Analytics

**Mini Project – Data Analysis using Python and Power BI**

Objective: clean, analyse and visualize IPL match data and prepare an interactive Power BI dashboard.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option('display.max_columns', None)

df = pd.read_csv('IPL_Cleaned.csv', parse_dates=['Match_Date'])
df.head()
```

## Part 1 – Data Loading

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
display(df.tail())
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()
print("Numerical columns:", numeric_cols)
print("Categorical/date columns:", categorical_cols)
```

## Part 2 – Data Exploration & Cleaning

```python
print(df.dtypes)
print("\nMissing values:")
display(df.isna().sum().sort_values(ascending=False).to_frame("Missing"))
print("\nDuplicate rows:", df.duplicated().sum())
df['Match_Date'] = pd.to_datetime(df['Match_Date'], errors='coerce')
print("Date range:", df['Match_Date'].min().date(), "to", df['Match_Date'].max().date())
```

```python
for c in ['Season','Winner','Toss_Decision','Result','Venue']:
    print(c, "unique values:", df[c].nunique())
```

## Part 3 – Data Analysis using Pandas & NumPy

```python
print("Total matches:", df['Match_ID'].nunique())
print("\nMatches by season:")
display(df.groupby('Season').size().rename('Matches').reset_index())

print("\nTeam wins:")
display(df['Winner'].value_counts().rename('Wins').reset_index())

print("Highest wins:", df['Winner'].value_counts().idxmax())
print("Lowest wins:", df['Winner'].value_counts().idxmin())
print("Most Player-of-the-Match awards:", df['Player_of_Match'].value_counts().idxmax())
print("Most frequent venue:", df['Venue'].value_counts().idxmax())
print("Toss winner match-win rate: {:.2f}%".format(df['Toss_Won_Match'].mean()*100))
print("Highest winning margin by runs:", df['Win_By_Runs'].max())
print("Highest winning margin by wickets:", df['Win_By_Wickets'].max())

print("\nTop 10 batsmen:")
display(df.groupby('Batsman')['Runs_Scored'].sum().sort_values(ascending=False).head(10))

print("\nTop 10 bowlers:")
display(df.groupby('Bowler')['Wickets'].sum().sort_values(ascending=False).head(10))

print("\nSeason-wise total runs:")
display(df.groupby('Season')['Total_Runs'].sum())

print("\nSeason-wise total wickets:")
display(df.groupby('Season')['Wickets'].sum())

print("\nAverage runs per match:", round(df['Total_Runs'].mean(),2))
print("\nTeam-wise average featured runs:")
display(df.groupby('Batting_Team')['Runs_Scored'].mean().sort_values(ascending=False))
```

## Part 4 – Python Data Visualization

```python
plots = [
    (df.groupby('Season').size(), 'bar', 'Matches Played by Season', 'Matches'),
    (df['Winner'].value_counts().sort_values(), 'barh', 'Matches Won by Team', 'Wins'),
    (df.groupby('Batsman')['Runs_Scored'].sum().sort_values(ascending=False).head(10).sort_values(), 'barh', 'Top 10 Run Scorers', 'Runs'),
    (df.groupby('Bowler')['Wickets'].sum().sort_values(ascending=False).head(10).sort_values(), 'barh', 'Top 10 Wicket Takers', 'Wickets'),
    (df['Player_of_Match'].value_counts().head(10).sort_values(), 'barh', 'Player of the Match Awards', 'Awards'),
    (df['Toss_Decision'].value_counts(), 'pie', 'Toss Decision Distribution', ''),
    (df.groupby('Season')['Total_Runs'].sum(), 'line', 'Season-wise Total Runs', 'Runs'),
    (df.groupby('Season')['Wickets'].sum(), 'line', 'Season-wise Total Wickets', 'Wickets'),
    (df['Venue'].value_counts().sort_values(), 'barh', 'Venue-wise Match Count', 'Matches')
]
for data, kind, title, ylabel in plots:
    plt.figure(figsize=(8,4.5))
    if kind == 'bar': data.plot(kind='bar')
    elif kind == 'barh': data.plot(kind='barh')
    elif kind == 'line': data.plot(marker='o')
    elif kind == 'pie': data.plot(kind='pie', autopct='%1.1f%%')
    plt.title(title)
    if ylabel: plt.ylabel(ylabel)
    if kind == 'pie': plt.ylabel('')
    plt.tight_layout()
    plt.show()

plt.figure(figsize=(7,5))
sns.heatmap(df[['Runs_Scored','Wickets','Extra_Runs','Total_Runs','Win_By_Runs','Win_By_Wickets','Toss_Won_Match']].corr(), annot=True, fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
